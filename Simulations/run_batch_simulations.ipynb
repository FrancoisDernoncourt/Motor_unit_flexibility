```python
import subprocess
import csv
import numpy as np
from itertools import product
import os
import pandas as pd
# requires nbconvert (pip install nbconvert)
```

```python
# Simulation parameters to try out
input_data_filename = ["\\S1_VL_HUFr_input_for_simulation.mat",
    "\\S2_VL_DEFr_input_for_simulation.mat",
    "\\S3_VL_TABa_input_for_simulation.mat",
    "\\S4_VL_MAVi_input_for_simulation.mat",
    "\\S5_VL_MOVa_input_for_simulation.mat",
    "\\S6_VL_DEPi_input_for_simulation.mat",
    "\\S8_VL_ZALo_input_for_simulation.mat"]

simulation_names = []
for input_data_filename_i in range(len(input_data_filename)):
    simulation_names.append(f"One_common_excitatory_input_S{input_data_filename_i+1}")

column_names = [
    'input_data_filename'
    ]

# Path where the notebooks are located
os.chdir('C:\\Users\\fdernoncourt\\Documents\\GitHub\\_New_flexibility_simulation\\')
```

```python
# Generate a list of all possible combinations
list_of_all_combinations = list(product(input_data_filename))
num_combinations = np.shape(list_of_all_combinations)[0]
df_combinations = pd.DataFrame(list_of_all_combinations, columns=column_names)
```

```python
print(f"Number of simulations = {num_combinations}")
print(f"Simmulation names = {simulation_names}")
```

```python
def run_simulation(param1, param_savename): # converting python to jupyter notebook files
    # Set environment variables for the parameters
    os.environ['input_data_filename'] = str(param1)
    os.environ['simulation_name'] = str(param_savename)
    
    # Run the Jupyter Notebook
    result = subprocess.run(
        ['python', '-m', 'nbconvert', '--to', 'notebook', '--execute', '--output',
         'executed_notebook_iteration_for_batch_simulation.ipynb',
         'Simulation_script.ipynb'], 
        capture_output=True, 
        text=True
    )
    if result.returncode != 0:
        print(f"Simulation failed with return code {result.returncode}")
        print("Error output:")
        print(result.stderr)
```

```python
# Run the simulations
for i in np.arange(num_combinations):
      print("Running simulation "+str(i+1)+"/"+str(num_combinations)+":\n"+\
      "     > "+simulation_names[i])
      result = run_simulation(df_combinations.loc[i, 'input_data_filename'], 
                            simulation_names[i])
```
